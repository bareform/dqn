## Deep Q-Learning Networks

```python
import os

import gymnasium as gym
import matplotlib.pyplot as plt
import torch
from IPython.display import HTML, display
from matplotlib.animation import FuncAnimation

from models import DQN
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")
```

```python
def load_checkpoint(checkpoint_path, device="cuda"):
    checkpoint = torch.load(checkpoint_path, map_location="cpu", weights_only=False)
    model = DQN(
        state_dim=checkpoint["state_dim"],
        action_dim=checkpoint["action_dim"],
        depth=checkpoint["depth"],
        W=checkpoint["W"],
    )
    model = model.to(device)
    model.load_state_dict(checkpoint["dqn"])
    return model
```

### FrozenLake-v1 4x4

```python
frozenlake_path = os.path.join("checkpoints", "frozenlake-v1-4x4_100000.pth")
```

```python
frozenlake_model = load_checkpoint(frozenlake_path)
```

```python
env = gym.make("FrozenLake-v1", map_name="4x4", is_slippery=False, render_mode="rgb_array")
state_dim = env.observation_space.n
action_dim = env.action_space.n

visualization_frames = []

state, _ = env.reset()
visualization_frames.append(env.render())
while True:
    state_tensor = torch.nn.functional.one_hot(
        torch.tensor(state, dtype=torch.long, device=device),
        num_classes=state_dim,
    ).float()
    with torch.inference_mode():
        action = int(frozenlake_model(state_tensor).argmax().item())

    state, _, terminated, truncated, _ = env.step(action)
    visualization_frames.append(env.render())
    if terminated or truncated:
        break

env.close()

fig, ax = plt.subplots(figsize=(4, 4))
image = ax.imshow(visualization_frames[0])
ax.axis("off")
def update_frame(frame):
    image.set_data(frame)
    return (image,)

animation = FuncAnimation(fig, update_frame, frames=visualization_frames, interval=500, blit=True)
plt.close(fig)
display(HTML(animation.to_jshtml()))
```

### Pendulum-v1

```python
pendulum_path = os.path.join("checkpoints", "pendulum-v1_300000.pth")
```

```python
pendulum_model = load_checkpoint(pendulum_path)
pendulum_checkpoint = torch.load(pendulum_path, map_location="cpu", weights_only=False)
pendulum_action_values = pendulum_checkpoint["action_values"]
```

```python
env = gym.make("Pendulum-v1", render_mode="rgb_array")
visualization_frames = []

state, _ = env.reset()
visualization_frames.append(env.render())
while True:
    state_tensor = torch.as_tensor(state, dtype=torch.float32, device=device)
    with torch.inference_mode():
        action_index = int(pendulum_model(state_tensor).argmax().item())

    action = env.action_space.low.copy()
    action[0] = pendulum_action_values[action_index]
    state, _, terminated, truncated, _ = env.step(action)
    visualization_frames.append(env.render())
    if terminated or truncated:
        break

env.close()

fig, ax = plt.subplots(figsize=(6, 4))
image = ax.imshow(visualization_frames[0])
ax.axis("off")
def update_frame(frame):
    image.set_data(frame)
    return (image,)

animation = FuncAnimation(fig, update_frame, frames=visualization_frames, interval=30, blit=True)
plt.close(fig)
display(HTML(animation.to_jshtml()))
```

### Blackjack-v1

```python
blackjack_path = os.path.join("checkpoints", "blackjack-v1_300000.pth")
```

```python
blackjack_model = load_checkpoint(blackjack_path)
```

```python
env = gym.make("Blackjack-v1", render_mode="rgb_array")
visualization_frames = []

state, _ = env.reset()
visualization_frames.append(env.render())
while True:
    player_sum, dealer_card, usable_ace = state
    state_tensor = torch.tensor(
        [player_sum / 31, dealer_card / 10, float(usable_ace)],
        dtype=torch.float32,
        device=device,
    )
    with torch.inference_mode():
        action = int(blackjack_model(state_tensor).argmax().item())

    state, _, terminated, truncated, _ = env.step(action)
    visualization_frames.append(env.render())
    if terminated or truncated:
        break

env.close()

fig, ax = plt.subplots(figsize=(6, 4))
image = ax.imshow(visualization_frames[0])
ax.axis("off")
def update_frame(frame):
    image.set_data(frame)
    return (image,)

animation = FuncAnimation(fig, update_frame, frames=visualization_frames, interval=250, blit=True)
plt.close(fig)
display(HTML(animation.to_jshtml()))
```
