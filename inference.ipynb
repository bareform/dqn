## Deep Q-Learning Networks

```python
import os
import random

import gymnasium as gym
import matplotlib.pyplot as plt
import torch
from IPython.display import HTML, display
from matplotlib.animation import FuncAnimation

from models import DQN
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")
```

```python
def load_checkpoint(checkpoint_path, device="cuda"):
    checkpoint = torch.load(checkpoint_path, map_location="cpu", weights_only=False)
    model = DQN(
        state_dim=checkpoint["state_dim"],
        action_dim=checkpoint["action_dim"],
        depth=checkpoint["depth"],
        W=checkpoint["W"],
    )
    model = model.to(device)
    model.load_state_dict(checkpoint["dqn"])
    return model
```

### FrozenLake-v1 4x4

```python
frozenlake_path = os.path.join("checkpoints", "frozenlake-v1-4x4_100000.pth")
```

```python
frozenlake_model = load_checkpoint(frozenlake_path)
```

```python
env = gym.make("FrozenLake-v1", map_name="4x4", is_slippery=False, render_mode="rgb_array")
state_dim = env.observation_space.n
action_dim = env.action_space.n

visualization_frames = []

state, _ = env.reset()
visualization_frames.append(env.render())
while True:
    state_tensor = torch.nn.functional.one_hot(
        torch.tensor(state, dtype=torch.long, device=device),
        num_classes=state_dim,
    ).float()
    with torch.inference_mode():
        action = int(frozenlake_model(state_tensor).argmax().item())

    state, _, terminated, truncated, _ = env.step(action)
    visualization_frames.append(env.render())
    if terminated or truncated:
        break

env.close()

fig, ax = plt.subplots(figsize=(4, 4))
image = ax.imshow(visualization_frames[0])
ax.axis("off")
def update_frame(frame):
    image.set_data(frame)
    return (image,)

animation = FuncAnimation(fig, update_frame, frames=visualization_frames, interval=500, blit=True)
plt.close(fig)
display(HTML(animation.to_jshtml()))
```
